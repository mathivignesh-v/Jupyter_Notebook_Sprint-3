## Copy vs View

### Step 1 – Understand the Concept

A view shares underlying data; a copy has independent data.

### Step 2 – Demonstrate the Concept

Understanding this prevents accidental changes during preprocessing.

### Step 3 – Implement the Concept

Implement the concept using Python and explain what the code does, why the function/method is used, and where it is useful in AI/ML or data analytics.

```python
import numpy as np
x=np.array([1,2,3])
v=x[:2]
c=x[:2].copy()
v[0]=99
c[1]=88
print('original:',x)
print('copy:',c)
```

### Interpretation

Changing the view changes the original; changing the copy does not.

### Key Takeaway

**Key Takeaway:** Changing the view changes the original; changing the copy does not.

## Best Practices
- Inspect data before transforming it.
- Prefer clear, readable code and meaningful names.
- Use vectorized NumPy/Pandas operations when appropriate.
- Preserve and validate data types.
- Check outputs after important transformations.
- Use reproducible random seeds for experiments.
- Avoid unintended mutation by using copies when necessary.

## Common Mistakes
- Confusing rows, columns, shape, and dimensions.
- Assuming every operation changes the original object.
- Ignoring missing values or incorrect dtypes.
- Using loops when vectorized operations are available.
- Loading or merging data without checking schemas and keys.
- Explaining syntax without understanding the output.

## Advantages
- NumPy provides efficient numerical arrays and vectorized computation.
- Pandas provides labeled tabular data manipulation.
- Both integrate with the wider Python AI/ML ecosystem.
- They support preprocessing, EDA, feature engineering, and data preparation.

## Limitations
- NumPy arrays work best with compatible numerical dtypes.
- Pandas can use significant memory for large datasets.
- Some convenient high-level operations can be expensive.
- Library functions do not replace domain knowledge or data validation.
